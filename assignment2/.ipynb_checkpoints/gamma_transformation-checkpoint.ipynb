## Practical 3: Gamma (Power-Law) Transformation

```python
import cv2
import numpy as np
```

```python
def cv2_cleanup():
    cv2.waitKey()
    cv2.destroyAllWindows()

def gamma_image(image, alpha):
    # s = alpha * r
    image = image.astype(np.float32)
    s = alpha * image
    s = np.array(s, dtype=np.uint8)
    return s

def show_gamma_transformation(image, alpha):
    cv2.imshow("Original", image)
    cv2.imshow("Gamma Transformation", gamma_image(image, alpha))
    cv2_cleanup()
```

```python
# Use Case 1: Correcting Underexposed Images

"""
A photograph captured in a poorly lit environment appears too dark. Apply Gamma 
Correction (γ < 1) to improve its brightness while preserving image details.
"""

image = cv2.imread("dark_bg.jpg")
image = cv2.resize(image, (500, 500))
show_gamma_transformation(image, 0.7)
```

```python
# Use Case 2: Correcting Overexposed Images

"""
An outdoor photograph captured under strong sunlight appears excessively bright. Apply 
Gamma Correction (γ > 1) to reduce brightness and improve visibility.
"""

image = cv2.imread("light_bg.jpg")
image = cv2.resize(image, (500, 500))
show_gamma_transformation(image, 1.25)
```

```python
# Use Case 3: Display Calibration

"""
Different display devices (monitors, TVs, projectors) reproduce brightness differently. Apply 
gamma correction with different gamma values (0.5, 1.0, 1.5, and 2.2) to observe their effect 
on image appearance.
"""

image = cv2.imread("standard_gamma.jpg")
image = cv2.resize(image, (500, 500))
for gamma in [0.5, 1.0, 1.5, 2.2]:
    show_gamma_transformation(image, gamma)
```
