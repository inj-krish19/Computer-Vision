# Practical 4: Comparison of Intensity Transformations

```python
import cv2
import numpy as np
```

```python
# Use Case: Image Enhancement for Different Applications
def cv2_cleanup():
    cv2.waitKey()
    cv2.destroyAllWindows()

def negative_transformation(image):
    return 255 - image

def log_transformation(image):
    # s = c * log(1 + r)
    image = image.astype(np.float32)
    c = 255 / np.log( 1+ np.max(image))
    s = c * (np.log(1 + image))
    s = np.array(s, dtype=np.uint8)
    return s

def gamma_transformation(image, alpha):
    # s = alpha * r
    image = image.astype(np.float32)
    s = alpha * image
    s = np.array(s, dtype=np.uint8)
    return s
```

```python
def intensity_transfomations(image):

    cv2.imshow("Original", image)
    cv2.imshow("Negation", negative_transformation(image))
    cv2.imshow("Log Transformation", log_transformation(image))
    cv2.imshow("Gamma 0.7 Transformation", gamma_transformation(image, 0.7))
    cv2.imshow("Gamma 1.2 Transformation", gamma_transformation(image, 1.2))
    cv2_cleanup()

paths = ["xray.jpg", "astronomical.jpg", "light_bg.jpg"]
for path in paths:
    image = cv2.imread(path)
    image = cv2.resize(image, (300, 300))
    intensity_transfomations(image)
```

# Transformations are best for:
- Medical X-ray images:  Image Negation, Gamma 0.7 Transformation
- Low-light photographs: Both Gamma Transformation
- Overexposed images:    Log Transformation, Gamma 0.7 Transformation
