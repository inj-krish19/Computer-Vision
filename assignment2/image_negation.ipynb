## Practical 1: Image Negation

```python
import cv2
import numpy as np
```

```python
def cv2_cleanup():
    cv2.waitKey()
    cv2.destroyAllWindows()

def show_negation_transformation(image):
    cv2.imshow("Original", image)
    cv2.imshow("Negative", 255 - image)
    cv2_cleanup()
```

```python
# Use Case 1: Medical Imaging (X-ray Analysis)

"""
Radiologists often examine the negative of an X-ray image because bones, tissues, and 
abnormalities may become more distinguishable in the inverted image. Write a program to 
read a grayscale X-ray image, generate its negative, and compare the original and negative 
images.
"""


image = cv2.imread("xray.jpg")
image = cv2.resize(image, (500, 500))
show_negation_transformation(image)
```

```python
# Use Case 2: Satellite Image Analysis

"""
A satellite image contains dark roads on a bright background. To improve visual 
interpretation, generate the negative of the image so that roads appear brighter than the 
surrounding area.
"""

image = cv2.imread("satellite.png")
image = cv2.resize(image, (500, 500))
show_negation_transformation(image)
```

```python
# Use Case 3: Historical Document Enhancement

"""
Old handwritten manuscripts often contain faded dark text on light paper. Generate the 
negative image to improve the visibility of the handwritten content.
"""

image = cv2.imread("document.jpg")
image = cv2.resize(image, (500, 500))
show_negation_transformation(image)
```
