## Practical 2: Log Transformation

```python
import cv2
import numpy as np
```

```python
def cv2_cleanup():
    cv2.waitKey()
    cv2.destroyAllWindows()

def log_image(image):
    # s = c * log(1 + r)
    image = image.astype(np.float32)
    c = 255 / np.log( 1 + np.max(image) )
    s = c * (np.log(1 + image))
    s = np.array(s, dtype=np.uint8)
    return s

def show_log_transformation(image):
    cv2.imshow("Original", image)
    cv2.imshow("Log Transformation", log_image(image))
    cv2_cleanup()
```

```python
# Use Case 1: Dark Image Enhancement

"""
A surveillance camera captures images under poor lighting conditions, making dark regions 
difficult to observe. Apply a logarithmic transformation to enhance the darker portions of the 
image while compressing brighter regions.
"""

image = cv2.imread("surveillance.jpg")
image = cv2.resize(image, (500, 500))
show_log_transformation(image)
```

```python
# Use Case 2: Astronomical Image Processing

"""
An astronomical image contains faint stars that are barely visible due to low intensity values. 
Apply logarithmic transformation to reveal the dim stars without excessively brightening 
already bright objects.
"""

image = cv2.imread("astronomical.jpg")
image = cv2.resize(image, (500, 500))
show_log_transformation(image)
```

```python
# Use Case 3: Medical Imaging

"""
A CT scan contains low-intensity anatomical structures that are difficult to identify. Enhance 
these structures using logarithmic transformation and compare the result with the original 
image.
"""

image = cv2.imread("ct_scan.jpg")
image = cv2.resize(image, (500, 500))
show_log_transformation(image)
```
