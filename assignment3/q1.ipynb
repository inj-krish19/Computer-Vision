```python
import cv2
import numpy as np
```

```python
# 1 - Write a Program for any image. Crop a particular portion of image. Change the intensity value of that particular section.

img = cv2.imread("image.jpg")
cv2.imshow("Image", img)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# middle center 100 pixels intensity values ae changed

cropped_img = img.copy()
height, width, _ = cropped_img.shape

mid_h, mid_w = height // 2, width // 2
cropped_img[mid_h-50:mid_h+50, mid_w-50:mid_w+50] = cropped_img[mid_h-50:mid_h+50, mid_w-50:mid_w+50] * 0.5

cv2.imshow("Changed Intensity Value Image", cropped_img)
cv2.waitKey()
cv2.destroyAllWindows()
```
