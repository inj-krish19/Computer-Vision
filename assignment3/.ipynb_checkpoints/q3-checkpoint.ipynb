```python
import numpy as np
```

```python
# 3 - Write a program to calculate all the three distance measures. Show that euclidean distance is minimum of all the three. 

def calculate_eucledian(a, b):
    if len(a) != len(b):
        return 0
    dims = len(a)
    return np.sqrt( sum( (a[i] - b[i]) ** 2 for i in range(dims) ))

def calculate_mahnhattan(a, b):
    if len(a) != len(b):
        return 0
    dims = len(a)
    return sum( abs(a[i] - b[i]) for i in range(dims))

def calculate_chebyshev(a, b):
    if len(a) != len(b):
        return 0
    dims = len(a)
    return max( abs(a[i] - b[i]) for i in range(dims) )

def calculate_minkowski(a, b, degree):
    if len(a) != len(b):
        return 0
    dims = len(a)
    return sum( abs(a[i] - b[i]) ** degree for i in range(dims) ) ** (1/degree)

print("Points a: (0, 0), b: (5, 5)")
print("Eucledian Distance of a and b is", calculate_eucledian((0, 0), (5, 5)) )
print("Manhattan Distance of a and b is", calculate_mahnhattan((0, 0), (5, 5)) )
print("Chebyshev Distance of a and b is", calculate_chebyshev((0, 0), (5, 5)) )
print("Minkowski Distance (degree=4) of a and b is", calculate_minkowski((0, 0), (5, 5), 4) )
```
