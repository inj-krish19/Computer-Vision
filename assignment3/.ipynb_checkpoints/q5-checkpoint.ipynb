```python
import cv2
import numpy as np
```

```python
# 5 - Perform Bit Plane Slicing

img = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)
cv2.imshow("Image", img)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# 7 bit plane slicing

for shift in range(8):
    bit_plane = np.array(((img >> shift) & 1) * 255, dtype=np.uint8)
    cv2.imshow(f"Bit Plane {shift}", bit_plane)
    cv2.waitKey()
    cv2.destroyAllWindows()
```
