```python
import cv2
import numpy as np
```

```python
# 2 - Enter any point in the image and mark all the 4 neighbours, 8 neighbours. 

img = cv2.imread("image.jpg")
cv2.imshow("Image", img)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# middle pixel's 4 neighbours

height, width, _ = img.shape

mh, mw = height // 2, width // 2
fourn = img.copy()

fourn[mh-1, mw] = (255, 0, 0)
fourn[mh, mw-1] = (255, 0, 0)
fourn[mh+1, mw] = (255, 0, 0)
fourn[mh, mw+1] = (255, 0, 0)

cv2.imshow("4 Neighbors", fourn)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# middle pixel's 4 neighbours

height, width, _ = img.shape

mh, mw = height // 2, width // 2
eightn = img.copy()

eightn[mh-1, mw] = (0, 0, 255)
eightn[mh, mw-1] = (0, 0, 255)
eightn[mh+1, mw] = (0, 0, 255)
eightn[mh, mw+1] = (0, 0, 255)

eightn[mh-1, mw-1] = (0, 0, 255)
eightn[mh-1, mw+1] = (0, 0, 255)
eightn[mh+1, mw-1] = (0, 0, 255)
eightn[mh+1, mw+1] = (0, 0, 255)

cv2.imshow("8 Neighbors", eightn)
cv2.waitKey()
cv2.destroyAllWindows()
```
