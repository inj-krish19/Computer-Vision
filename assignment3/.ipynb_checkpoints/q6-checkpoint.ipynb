```python
import cv2
import numpy as np
```

```python
def cv2_cleanup():
    cv2.waitKey()
    cv2.destroyAllWindows()

img = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)
cv2.imshow("Image", img)
cv2_cleanup()
```

```python
# 6 - Program to perform image smoothening using filters of different sizes with hard coding as well as opencv

# Mean Filter

def mean_filter(image, size):
    height, width = image.shape
    fh, fw = size

    padded_img = np.pad(img, size, mode='edge')
    result = np.zeros((height, width), dtype=np.float32)

    for i in range(height):
        for j in range(width):
            window = padded_img[i:i+fh, j:j+fw]
            result[i, j] = np.mean(window)

    return np.uint8(result)

cv2.imshow("Custom - Mean Filter 3x3", mean_filter(img, (3, 3)))
cv2.imshow("Custom - Mean Filter 5x5", mean_filter(img, (5, 5)))
cv2.imshow("Custom - Mean Filter 7x7", mean_filter(img, (7, 7)))
cv2_cleanup()

cv2.imshow("Builtin - Mean Filter 3x3", cv2.blur(img, (3, 3)))
cv2.imshow("Builtin - Mean Filter 5x5", cv2.blur(img, (5, 5)))
cv2.imshow("Builtin - Mean Filter 7x7", cv2.blur(img, (7, 7)))
cv2_cleanup()
```

```python
# Gaussian Filter

def gaussian_kernel(size, sigma=1):
    fh, fw = size
    ax = np.arange(-fh // 2 + 1, fw // 2 + 1)
    xx, yy = np.meshgrid(ax, ax)
    
    kernel = np.exp(-(xx**2 + yy**2) / (2 * sigma**2))
    kernel = kernel / np.sum(kernel)
    return kernel * 16
    
def gaussian_filter(image, size):
    height, width = image.shape
    fh, fw = size

    padded_img = np.pad(img, size, mode='edge')
    result = np.zeros((height, width), dtype=np.float32)
    
    gaussian_filter = gaussian_kernel(size, 1)
    for i in range(height):
        for j in range(width):
            window = padded_img[i:i+fh, j:j+fw]
            result[i, j] = np.mean(window * gaussian_filter)

    return np.uint8(result)

cv2.imshow("Custom - Gaussian Filter 3x3", gaussian_filter(img, (3, 3)))
cv2.imshow("Custom - Gaussian Filter 5x5", gaussian_filter(img, (5, 5)))
cv2.imshow("Custom - Gaussian Filter 7x7", gaussian_filter(img, (7, 7)))
cv2_cleanup()

cv2.imshow("Builtin - Gaussian Filter 3x3", cv2.GaussianBlur(img, (3, 3), 1))
cv2.imshow("Builtin - Gaussian Filter 5x5", cv2.GaussianBlur(img, (5, 5), 1))
cv2.imshow("Builtin - Gaussian Filter 7x7", cv2.GaussianBlur(img, (7, 7), 1))
cv2_cleanup()
```

```python
# Median Filter

def median_filter(image, size):
    height, width = image.shape
    fw, fh = size

    padded_img = np.pad(image, size, mode='edge')
    result = np.zeros((height, width), dtype=np.float32)
    for i in range(height):
        for j in range(width):
            window = padded_img[i:i+fh, j:j+fw]
            result[i, j] = np.median(window)

    return np.uint8(result)

cv2.imshow("Custom - Median Filter 3x3", median_filter(img, (3, 3)))
cv2.imshow("Custom - Median Filter 5x5", median_filter(img, (5, 5)))
cv2.imshow("Custom - Median Filter 7x7", median_filter(img, (7, 7)))
cv2_cleanup()

cv2.imshow("Builtin - Median Filter 3x3", cv2.medianBlur(img, 3))
cv2.imshow("Builtin - Median Filter 5x5", cv2.medianBlur(img, 5))
cv2.imshow("Builtin - Median Filter 7x7", cv2.medianBlur(img, 7))
cv2_cleanup()
```
