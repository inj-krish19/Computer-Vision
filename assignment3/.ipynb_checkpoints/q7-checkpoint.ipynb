```python
import cv2
import numpy as np
```

```python
def cv2_cleanup():
    cv2.waitKey()
    cv2.destroyAllWindows()

img = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)
cv2.imshow("Image", img)
cv2_cleanup()
```

```python
# 7 - Write hard coded program for 2D convolution for image averaging (simple and weighted) and median. 

def simple_average(img):
    height, width = img.shape
    padded_img = np.pad(img, (3, 3), mode='edge')

    result = np.zeros((height, width))
    for i in range(height):
        for j in range(width):
            result[i, j] = np.mean(padded_img[i:i+3, j:j+3])

    return np.uint8(result)


def weighted_average(img):
    height, width = img.shape
    padded_img = np.pad(img, (3, 3), mode='edge')

    kernel = np.array([[1, 2, 1], [2, 4, 2], [1, 2, 1]])
    result = np.zeros((height, width))
    for i in range(height):
        for j in range(width):
            result[i, j] = (1/ 16) * np.sum(padded_img[i:i+3, j:j+3] * kernel)

    return np.uint8(result)


def median_average(img):
    height, width = img.shape
    padded_img = np.pad(img, (3, 3), mode='edge')

    result = np.zeros((height, width))
    for i in range(height):
        for j in range(width):
            result[i, j] = np.median(padded_img[i:i+3, j:j+3])

    return np.uint8(result)

cv2.imshow("Simple Average", simple_average(img))
cv2.imshow("Weighted Average", weighted_average(img))
cv2.imshow("Median Average", median_average(img))
cv2_cleanup()
```
