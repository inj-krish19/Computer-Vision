```python
import cv2
import numpy as np
```

```python
def cv2_cleanup():
    cv2.waitKey()
    cv2.destroyAllWindows()

img = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)
cv2.imshow("Image", img)
cv2_cleanup()
```

```python
# 9 - Write a program to detect edges in an image using OpenCV: 
# a) Filter2D with Laplacian Kernel 
# b) Laplacian Method in OpenCV 
# c) Sobel Operator

# a - Laplacin Kernel with Filter 2D
laplacian_kernel = np.array([[0, 1, 0], [1, -4, 1], [0, 1, 0]])
laplacian_filter = cv2.filter2D(img, cv2.CV_64F, laplacian_kernel)
laplacian_filter = cv2.convertScaleAbs(laplacian_filter)

cv2.imshow("Laplcian Kernel with Filter 2D", laplacian_filter)
cv2_cleanup()
```

```python
# b - Laplacin Kernel with OpenCV
laplacian = cv2.Laplacian(img, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)

cv2.imshow("Laplcian Kernel with OpenCV", laplacian)
cv2_cleanup()
```

```python
# b - Sobel Operator
sobel_x = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)
sobel_x = cv2.convertScaleAbs(sobel_x)
sobel_y = cv2.convertScaleAbs(sobel_y)
sobel = cv2.addWeighted(sobel_x, 0.5, sobel_y, 0.5, 0)

cv2.imshow("Sobel Operator", sobel)
cv2_cleanup()
```
