```python
import cv2
import numpy as np
```

```python
def cv2_cleanup():
    cv2.waitKey()
    cv2.destroyAllWindows()

img = cv2.imread("image.jpg", cv2.IMREAD_GRAYSCALE)
cv2.imshow("Image", img)
cv2_cleanup()
```

```python
# 8 - Perform image smoothening (noise reduction) using: 
# a) Mean Filter(Average Filter) 
# b) Gaussian Blur 
# c) Median Filter 
# d) Weighted Mean Filter 
# e) Min and Max Filters 

# a - Mean Filter
cv2.imshow("Mean Image Smoothning", cv2.blur(img, (5, 5)))
cv2_cleanup()
```

```python
# b - Guassian Blur
cv2.imshow("Gaussian Image Smoothning", cv2.GaussianBlur(img, (5, 5), 0))
cv2_cleanup()
```

```python
# c - Median Filter
cv2.imshow("Median Image Smoothning", cv2.medianBlur(img, 5))
cv2_cleanup()
```

```python
# d - Weighted Mean Filter
kernel = np.array([[1, 2, 1], [2, 4, 2], [1, 2, 1]])
kernel = kernel / np.sum(kernel)

cv2.imshow("Weighted Mean Image Smoothning", cv2.filter2D(img, -1, kernel))
cv2_cleanup()
```

```python
# e - Min Filter
cv2.imshow("Min Image Smoothning", cv2.erode(img, np.ones((5, 5))))
cv2_cleanup()
```

```python
# e - Max Filter
cv2.imshow("Max Image Smoothning", cv2.dilate(img, np.ones((5, 5))))
cv2_cleanup()
```
