```python
import numpy as np
import cv2
```

```python
img = cv2.imread("image2.jpg")
height, width, channels = img.shape
height, width, channels
```

```python
def render_image(label, array):
    # cv2.namedWindow(label, cv2.WINDOW_NORMAL)
    cv2.imshow(label, array)
    # cv2.resizeWindow(label, 800, 400)
    cv2.waitKey()
    cv2.destroyAllWindows()

render_image("Original", img)
```

```python
# 1 - Swap the left and right halves of an image. 

# for swapping on horizontal cut for columns
mid_w = width // 2
right_left = np.hstack((img[:, mid_w:, :], img[:, :mid_w, :]))
render_image("Right Left", right_left)
```

```python
# 2 - Swap the top and bottom halves of an image.

# for swapping on vertical cut for rows
mid_h = height // 2
bottom_top = np.vstack((img[mid_h:, :, :], img[:mid_h, :, :]))
render_image("Bottom Top", bottom_top) 
```

```python
# 3- Create a horizontal mirror image using NumPy slicing. 

# for mirror instead of 0...n go n...0 for columns
# carefull: image of both nin looks natural but it happens
mirror_horizontal = img[:, ::-1, :].copy()
render_image("Mirror Horizontal", mirror_horizontal)
```

```python
# 4 - Create a vertical mirror image using NumPy slicing. 

mirror_vertical = img[::-1, :, :]
render_image("Mirror Vertical", mirror_vertical)
```

```python
# 5 - Divide an image into 9 equal blocks and display each block separately. 

# don't run else you have time to close it
block_number = 1
difference_h, difference_w = height // 3, width // 3

for h in range(0, height, difference_h):
    for w in range(0, width, difference_w):
        render_image(f"Block {block_number}", img[h:h+difference_h, w:w+difference_w, :])
        block_number = block_number + 1
```

```python
# 6 - Display only the Blue, Green, and Red channels as separate grayscale images. 

# opencv follows BGR not RGB
colors = ["Blue", "Green", "Red"]
for index in range(len(colors)):
    render_image(colors[index], img[:, :, index])
```

```python
# 7 - Crop the center 200 × 200 region, resize it to 600 × 600, and display both images. 

center_h, center_w = height // 2, width // 2
center_200 = img[center_h-100:center_h+101, center_w-100:center_w+101, :]

# 200 x 200 Region Rendering - No Resizing
cv2.imshow("Center 200", center_200)
cv2.waitKey()
cv2.destroyAllWindows()

# 200 x 200 Region Rendering - 600 x 600 Size
cv2.namedWindow("Center 200", cv2.WINDOW_NORMAL)
cv2.imshow("Center 200", center_200)
cv2.resizeWindow("Center 200", 600, 600)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# 8 - Resize an image to 25%, 50%, and 200% of its original size and compare the outputs.

# i) - 25%
quater = cv2.resize(img, (int(width * 0.25), int(height * 0.25)), interpolation=cv2.INTER_AREA)
cv2.imshow("25%", quater)
cv2.waitKey()
cv2.destroyAllWindows()

# ii) - 50%
half = cv2.resize(img, (int(width * 0.5), int(height * 0.5)), interpolation=cv2.INTER_AREA)
cv2.imshow("50%", half)
cv2.waitKey()
cv2.destroyAllWindows()

# iii) - 200%
double = cv2.resize(img, (int(width * 2), int(height * 2)), interpolation=cv2.INTER_CUBIC)
cv2.imshow("200%", double)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# 9 - Change an entire column to Green. 

# made column 100 as green
green_color = np.array([
    [(0, 255, 0)  for _ in range(height) ]
])
green_column = img.copy()
green_column[:, 100, :] = green_color
render_image("Green Column", green_column)
```

```python
# 10 - Change an entire row to Blue.

# made row 100 as blue
blue_color = np.array([
    [(255, 0, 0) for _ in range(width)]
])
blue_row = img.copy()
blue_row[100, :, :] = blue_color
render_image("Blue Row", blue_row)
```

```python
# 11 - Draw a red border around the image using pixel manipulation. 

red_row_border = np.array([
    [(0, 0, 255) for _ in range(width)]
])

red_column_border = np.array([
    [(0, 0, 255) for _ in range(height)]
])

red_border = img.copy()

red_border[0, :, :] = red_row_border
red_border[-1, :, :] = red_row_border
red_border[:, 0, :] = red_column_border
red_border[:, -1, :] = red_column_border

cv2.namedWindow("Red Border", cv2.WINDOW_NORMAL)
cv2.imshow("Red Border", red_border)
cv2.resizeWindow("Red Border", 800, 600)
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
# 12 - Crop the bottom-right 250 × 250 region. 

bottom_right_250 = img[-250:, -250:, :].copy()
render_image("Bottom Right 250", bottom_right_250)
```

```python
# 13 - Crop the top-left 200 × 200 region of an image. 

top_left_200 = img[:200, :200, :].copy()
render_image("Top Left 200", top_left_200)
```

```python
# 14 - Draw a green square using pixel modification. 

# middle 100 x 100 are green
green_region = np.array([
    [(0, 255, 0) for _ in range(100)] for _ in range(100)
])
green_square = img.copy()
green_square[mid_h-50:mid_h+50 , mid_w-50:mid_w+50 ,:] = green_region
render_image("Green Square", green_square)
```

```python
# 15 - Crop all four quadrants of an image. 

quadrant = 1
for h in range(0, height, mid_h):
    for w in range(0, height, mid_w):
        render_image(f"Quadrants {quadrant}", img[h:h+mid_h, w:w+mid_w, :])
        quadrant += 1
```
